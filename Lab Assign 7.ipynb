```python
from cryptography.hazmat.primitives.asymmetric import rsa
from cryptography.hazmat.primitives import serialization

# Generate keys
private_key = rsa.generate_private_key(
    public_exponent=65537,
    key_size=2048
)

public_key = private_key.public_key()

# Save private key
with open("private.pem", "wb") as f:
    f.write(private_key.private_bytes(
        encoding=serialization.Encoding.PEM,
        format=serialization.PrivateFormat.PKCS8,
        encryption_algorithm=serialization.NoEncryption()
    ))

# Save public key
with open("public.pem", "wb") as f:
    f.write(public_key.public_bytes(
        encoding=serialization.Encoding.PEM,
        format=serialization.PublicFormat.SubjectPublicKeyInfo
    ))

print("RSA Keys Generated")
```

```python
from cryptography.hazmat.primitives.asymmetric import padding
from cryptography.hazmat.primitives import hashes

# Load public key
with open("public.pem", "rb") as f:
    public_key = serialization.load_pem_public_key(f.read())

# Read file
with open("data.txt", "rb") as file:
    data = file.read()

# Encrypt
encrypted = public_key.encrypt(
    data,
    padding.OAEP(
        mgf=padding.MGF1(algorithm=hashes.SHA256()),
        algorithm=hashes.SHA256(),
        label=None
    )
)

with open("rsa_encrypted.txt", "wb") as f:
    f.write(encrypted)

print("RSA Encryption Done")
```

```python
# Load private key
with open("private.pem", "rb") as f:
    private_key = serialization.load_pem_private_key(f.read(), password=None)

# Read encrypted file
with open("rsa_encrypted.txt", "rb") as f:
    encrypted = f.read()

# Decrypt
decrypted = private_key.decrypt(
    encrypted,
    padding.OAEP(
        mgf=padding.MGF1(algorithm=hashes.SHA256()),
        algorithm=hashes.SHA256(),
        label=None
    )
)

with open("rsa_decrypted.txt", "wb") as f:
    f.write(decrypted)

print("RSA Decryption Done")
```

```python
import time
from cryptography.fernet import Fernet
from cryptography.hazmat.primitives.asymmetric import padding
from cryptography.hazmat.primitives import hashes
from cryptography.hazmat.primitives import serialization

# Re-initialize Fernet object 'f' and ensure other variables are loaded
# Load key
with open("secret.key", "rb") as key_file:
    key = key_file.read()
f = Fernet(key)

# Read file data again
with open("data.txt", "rb") as file_data:
    data = file_data.read()

# Load public key again (using a different file variable name to avoid conflict)
with open("public.pem", "rb") as public_key_file:
    public_key = serialization.load_pem_public_key(public_key_file.read())

# Symmetric timing
start = time.time()
f.encrypt(data)
end = time.time()
print("Fernet Time:", end - start)

# RSA timing
start = time.time()
public_key.encrypt(
    data,
    padding.OAEP(
        mgf=padding.MGF1(algorithm=hashes.SHA256()),
        algorithm=hashes.SHA256(),
        label=None
    )
)
end = time.time()
print("RSA Time:", end - start)
```
