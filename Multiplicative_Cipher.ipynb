```python
# Multiplicative Cipher Implementation
# simple version (lab style)

import math

alphabet = "ABCDEFGHIJKLMNOPQRSTUVWXYZ"


# check if key is valid
def valid_key(k):
    if math.gcd(k, 26) == 1:
        return True
    return False


# modular inverse (simple way)
def mod_inv(k):
    for i in range(1, 26):
        if (k * i) % 26 == 1:
            return i
    return None


# encryption function
def encrypt(text, key):
    text = text.upper()
    result = ""

    for ch in text:
        if ch in alphabet:
            p = alphabet.index(ch)
            c = (p * key) % 26
            result += alphabet[c]
        else:
            result += ch   # keep spaces or symbols

    return result


# decryption function
def decrypt(text, key):
    text = text.upper()
    result = ""

    inv = mod_inv(key)

    if inv is None:
        print("No inverse found")
        return ""

    for ch in text:
        if ch in alphabet:
            c = alphabet.index(ch)
            p = (c * inv) % 26
            result += alphabet[p]
        else:
            result += ch

    return result


# brute force attack
def attack(cipher):
    print("\nTrying all possible keys...\n")

    for k in range(1, 26):
        if valid_key(k):
            plain = decrypt(cipher, k)
            print("Key =", k, "->", plain)


# ---------------- main ----------------

print("Multiplicative Cipher Program")

print("1. Encrypt")
print("2. Decrypt")
print("3. Cryptanalysis")

choice = input("Enter choice: ")

if choice == "1":
    msg = input("Enter message: ")
    key = int(input("Enter key: "))

    if valid_key(key):
        print("Encrypted:", encrypt(msg, key))
    else:
        print("Invalid key! Use numbers coprime with 26")

elif choice == "2":
    msg = input("Enter ciphertext: ")
    key = int(input("Enter key: "))

    if valid_key(key):
        print("Decrypted:", decrypt(msg, key))
    else:
        print("Invalid key!")

elif choice == "3":
    msg = input("Enter ciphertext: ")
    attack(msg)

else:
    print("Wrong choice")
```
