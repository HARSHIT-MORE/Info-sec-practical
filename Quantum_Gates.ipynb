```python
# Install Qiskit first (run once if not installed)
!pip install qiskit qiskit-aer

from qiskit import QuantumCircuit, transpile
from qiskit_aer import AerSimulator
from qiskit.visualization import plot_histogram
import matplotlib.pyplot as plt

# Create simulator
simulator = AerSimulator()
shots = 1024

# ---------------------------------------------------
# 1. Single Qubit Circuit with X Gate
# ---------------------------------------------------
qc1 = QuantumCircuit(1, 1)
qc1.x(0)                 # Apply X gate
qc1.measure(0, 0)        # Measure

print("1. Circuit with X Gate:")
print(qc1.draw())

compiled1 = transpile(qc1, simulator);
result1 = simulator.run(compiled1, shots=shots).result()
counts1 = result1.get_counts()

print("Measurement Results:", counts1)

plt.figure()
plot_histogram(counts1)
plt.title("Histogram - X Gate")
plt.show()


# ---------------------------------------------------
# 2. Single Qubit Circuit with H Gate
# ---------------------------------------------------
qc2 = QuantumCircuit(1, 1)
qc2.h(0)                 # Apply H gate
qc2.measure(0, 0)

print("\n2. Circuit with H Gate:")
print(qc2.draw())

compiled2 = transpile(qc2, simulator)
result2 = simulator.run(compiled2, shots=shots).result()
counts2 = result2.get_counts()

print("Measurement Results:", counts2)

plt.figure()
plot_histogram(counts2)
plt.title("Histogram - H Gate")
plt.show()


# ---------------------------------------------------
# 3. Two-Qubit Circuit with CNOT Gate
# ---------------------------------------------------
qc3 = QuantumCircuit(2, 2)
qc3.h(0)                 # Put control qubit in superposition
qc3.cx(0, 1)             # Apply CNOT
qc3.measure([0, 1], [0, 1])

print("\n3. Circuit with CNOT Gate:")
print(qc3.draw())

compiled3 = transpile(qc3, simulator)
result3 = simulator.run(compiled3, shots=shots).result()
counts3 = result3.get_counts()

print("Measurement Results:", counts3)

plt.figure()
plot_histogram(counts3)
plt.title("Histogram - CNOT Gate")
plt.show()
```

```python
from qiskit import QuantumCircuit, transpile
from qiskit_aer import AerSimulator
from qiskit.visualization import plot_histogram
import matplotlib.pyplot as plt

# Create simulator
simulator = AerSimulator()
shots = 1024
```
