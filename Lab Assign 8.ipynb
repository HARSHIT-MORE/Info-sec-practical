```python
with open("file.txt", "w") as f:
    f.write("This is a secure file.")
```

```python
import hashlib

with open("file.txt", "rb") as f:
    data = f.read()

hash_val = hashlib.sha512(data).hexdigest()

print("SHA-512 Hash:")
print(hash_val)
```

```python
import hmac

key = b'secret_key'

hmac_val = hmac.new(key, data, hashlib.sha512).hexdigest()

print("\nHMAC (SHA-512):")
print(hmac_val)
```

```python
with open("file.txt", "a") as f:
    f.write("\nFile has been modified!")
```

```python
with open("file.txt", "rb") as f:
    new_data = f.read()

new_hash = hashlib.sha512(new_data).hexdigest()
new_hmac = hmac.new(key, new_data, hashlib.sha512).hexdigest()

print("\nNew SHA-512 Hash:")
print(new_hash)

print("\nNew HMAC:")
print(new_hmac)
```

```python
print("\nComparison Result:")

if hash_val == new_hash:
    print("Hash unchanged")
else:
    print("Hash changed → File Tampered")

if hmac_val == new_hmac:
    print("HMAC unchanged")
else:
    print("HMAC changed → File Tampered")
```

```python
import hashlib

def get_hash(file):
    with open(file, 'rb') as f:
        return hashlib.sha512(f.read()).hexdigest()
```

```python
original_hash = get_hash("file.txt") #STORE Original Hash
```

```python
current_hash = get_hash("file.txt")
```

```python
if original_hash == current_hash:
    print("Integrity Verified")
else:
    print("Integrity Compromised")
```
