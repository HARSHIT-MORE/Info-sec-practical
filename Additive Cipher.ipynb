```python
def encrypt(text, key):
    result = ""
    for char in text:
        if char.isalpha():
            shift = ord('A') if char.isupper() else ord('a')
            result += chr((ord(char) - shift + key) % 26 + shift)
        else:
            result += char
    return result


def decrypt(text, key):
    result = ""
    for char in text:
        if char.isalpha():
            shift = ord('A') if char.isupper() else ord('a')
            result += chr((ord(char) - shift - key) % 26 + shift)
        else:
            result += char
    return result


# --- User Input ---
plaintext = input("Enter plaintext: ")
key = int(input("Enter key (0-25): "))

ciphertext = encrypt(plaintext, key)
decrypted = decrypt(ciphertext, key)

print("\nPlaintext :", plaintext)
print("Ciphertext:", ciphertext)
print("Decrypted :", decrypted)
```

```python
def brute_force(ciphertext):
    results = {}
    print("\nBrute Force Results:\n")
    for key in range(26):
        decrypted = decrypt(ciphertext, key)
        print(f"Key {key}: {decrypted}")
        results[key] = decrypted
    return results


# Example usage
cipher_input = input("\nEnter ciphertext for brute force: ")
brute_force(cipher_input)
```

```python
def frequency_analysis(ciphertext):
    freq = {}
    predicted_keys = []

    # Step 1: Count frequency manually
    for char in ciphertext:
        if char.isalpha():
            char = char.lower()
            if char in freq:
                freq[char] += 1
            else:
                freq[char] = 1

    # Step 2: Display frequencies
    print("\nLetter Frequencies:")
    for char in freq:
        print(char, ":", freq[char])

    # Step 3: Find most frequent letter
    max_char = None
    max_count = 0

    for char in freq:
        if freq[char] > max_count:
            max_count = freq[char]
            max_char = char

    print("\nMost frequent letter:", max_char)

    # Step 4: Assume mapping to common English letters
    common_letters = ['e', 't', 'a']

    print("\nPossible keys and decryptions:\n")
    if max_char:
        for letter in common_letters:
            key = (ord(max_char) - ord(letter)) % 26
            print("Assuming", max_char, "->", letter, "| Key =", key)
            print("Decryption:", decrypt(ciphertext, key))
            predicted_keys.append(key)
            print()
    return predicted_keys
```

```python
frequency_analysis("KHOOR")
```

```python
def comparative_analysis(ciphertext):
    print("\n--- Brute Force Results ---")
    brute_results = brute_force(ciphertext)

    for key, text in brute_results.items():
        print(f"Key {key:2}: {text}")

    print("\n--- Frequency Analysis ---")
    freq_keys = frequency_analysis(ciphertext)

    print("Predicted Keys:", freq_keys)

    print("\n--- Comparison ---")
    for key in freq_keys:
        print(f"Key {key} gives: {decrypt(ciphertext, key)}")

    print("\nConclusion:")
    print("Brute force shows all possibilities (100% accurate).")
    print("Frequency analysis predicts likely keys but may not always be correct.")
```

```python
comparative_analysis("KHOOR")
```
