```python
def encrypt(text, key):
    text = text.replace(" ", "").upper()
    col = len(key)

    # Padding with X
    while len(text) % col != 0:
        text += 'X'

    # Create matrix
    matrix = []
    for i in range(0, len(text), col):
        matrix.append(list(text[i:i+col]))

    # Get column order
    order = sorted(range(len(key)), key=lambda x: key[x])

    # Read columns
    cipher = ""
    for i in order:
        for row in matrix:
            cipher += row[i]

    return cipher


def decrypt(cipher, key):
    col = len(key)
    row = len(cipher) // col

    # Get column order
    order = sorted(range(len(key)), key=lambda x: key[x])

    # Create empty matrix
    matrix = [['' for _ in range(col)] for _ in range(row)]

    index = 0
    # Fill columns
    for i in order:
        for j in range(row):
            matrix[j][i] = cipher[index]
            index += 1

    # Read rows
    text = ""
    for row in matrix:
        text += ''.join(row)

    return text.rstrip('X')


# ---- MAIN ----
plaintext = input("Enter plaintext: ")
key = input("Enter key: ")

cipher = encrypt(plaintext, key)
print("Ciphertext:", cipher)

original = decrypt(cipher, key)
print("Decrypted text:", original)
```
